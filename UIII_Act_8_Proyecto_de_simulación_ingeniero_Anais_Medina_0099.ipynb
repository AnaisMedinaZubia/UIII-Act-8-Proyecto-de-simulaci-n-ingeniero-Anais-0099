```python
# --- SISTEMA DE IA: ASISTENTE DE SEGURIDAD PARA INGENIEROS ---
# Simulación interactiva para gestión de cascos de seguridad.

opcion = "0"

# Ciclo principal para mantener el menú activo [5.2, 5.8.1]
while opcion != "4":
    print("\n========================================")
    print("   CENTRO DE MANDO: SEGURIDAD (IA)   ")
    print("========================================")
    print("1. Identificar color de casco por cargo")
    print("2. Diagnóstico de integridad del casco")
    print("3. Calcular riesgo de impacto (IA)")
    print("4. Salir del sistema")

    opcion = input("\nSeleccione una opción (1-4): ") # Captura de datos [3, 4]

    # OPCIÓN 1: Clasificación de personal por color de casco [5, 6]
    if opcion == "1":
        print("\n[IA] Iniciando identificación de rol...")
        print("Cargos: 1. Ingeniero/Jefe, 2. Operario, 3. Electricista, 4. Visitante")
        cargo = input("Ingrese el número del cargo: ")

        if cargo == "1":
            print("RECOMENDACIÓN: El Ingeniero debe portar casco BLANCO.")
        elif cargo == "2":
            print("RECOMENDACIÓN: El Operario debe portar casco AMARILLO.")
        elif cargo == "3":
            print("RECOMENDACIÓN: El Electricista debe portar casco AZUL.")
        elif cargo == "4":
            print("RECOMENDACIÓN: El Visitante debe portar casco GRIS.")
        else:
            print("ALERTA: Cargo no identificado. Consulte manual de obra.")

    # OPCIÓN 2: Evaluación de sensores de equipo (Aritmética y Relacional) [4.56, 3.7.1]
    elif opcion == "2":
        print("\n[IA] Realizando escaneo de integridad...")
        antiguedad = int(input("¿Cuántos años de uso tiene el casco?: "))
        impactos = int(input("¿Cuántos impactos previos ha recibido?: "))
        fisuras = input("¿Presenta fisuras visibles? (si/no): ").lower()

        # Lógica de seguridad combinada con operadores lógicos [4.60, 4.68]
        if antiguedad > 5 or impactos >= 1 or fisuras == "si":
            print("DIAGNÓSTICO: ¡CASCO NO SEGURO! Proceder al reemplazo inmediato.")
        elif antiguedad > 3 and antiguedad <= 5:
            print("DIAGNÓSTICO: PRECAUCIÓN. El casco está cerca de su fecha de caducidad.")
        else:
            print("DIAGNÓSTICO: ESTADO ÓPTIMO. El equipo cumple con las normas.")

    # OPCIÓN 3: Cálculo de riesgo con operadores aritméticos [4.42, 4.43]
    elif opcion == "3":
        print("\n[IA] Calculando fuerza de impacto potencial...")
        peso_objeto = float(input("Ingrese peso del objeto que podría caer (kg): "))
        altura_caida = float(input("Ingrese altura estimada de la caída (metros): "))

        # Fórmula simplificada de energía potencial: Masa * Gravedad (9.8) * Altura [3.7.1]
        energia = peso_objeto * 9.8 * altura_caida

        print(f"La energía potencial de impacto es: {energia:.2f} Julios.")

        # Evaluación del nivel de peligro mediante condicionales [7, 8]
        if energia > 100:
            print("RIESGO: CRÍTICO. Se requiere casco de Clase G o E de alta resistencia.")
        else:
            print("RIESGO: MODERADO. El casco estándar es suficiente para este entorno.")

    # OPCIÓN 4: Salida del programa [5.9]
    elif opcion == "4":
        print("\nCerrando sistema de seguridad. ¡Trabaje con cuidado, ingeniero!")

    # Manejo de entradas no válidas en el menú [9]
    else:
        print("\n[Error] Opción no válida. Intente de nuevo.")
        print("Programa realizado por Anais Medina NL = 40")

print("\n--- Sistema fuera de línea ---")
```
